```python
# ## GÖREV: 
# Özellik Seçimi: Hangi özelliklerin kümeleme için kullanılacağına karar verin.  
# Küme Sayısını Belirleme: Elbow yöntemi gibi tekniklerle optimal küme sayısını belirleyin.  
# Kümeleme İşlemi: K-Means algoritmasını kullanarak verileri kümeleyin.  
# Sonuçları Görselleştirme: Kümeleme sonuçlarını uygun grafiklerle görselleştirin ve yorumlayın.
```

```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("dava.csv")
```

```python
# Özellik Seçimi: Hangi özelliklerin kümeleme için kullanılacağına karar verin.  
```

```python
X_basit = df[['Case Duration (Days)', 'Legal Fees (USD)']].values

# 2. Kümeleme İşlemi (K=3 ile Basit Yöntem)
# n_init=10 ile en iyi sonucu bulmayı garanti ediyoruz
kmeans_basit = KMeans(n_clusters=3, init='k-means++', random_state=42, n_init=10)
kmeans_basit.fit(X_basit)

# 3. Küme Sonuçlarını DataFrame'e Ekleme
df['Cluster'] = kmeans_basit.labels_

# 4. Sonuçları Görselleştirme
plt.figure(figsize=(9, 6))
plt.scatter(df['Case Duration (Days)'], df['Legal Fees (USD)'], 
            c=df['Cluster'], cmap='viridis', s=50, alpha=0.7)

# Küme merkezlerini çizme (K-Means'in bulduğu grupların ortası)
centers = kmeans_basit.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], 
            marker='X', s=200, color='red', label='Küme Merkezleri')

plt.title('Basit K-Means Kümeleme (K=3)')
plt.xlabel('Dava Süresi (Gün)')
plt.ylabel('Hukuk Maliyetleri (USD)')
plt.legend()
plt.grid(True)
plt.show()

# Küme merkezlerini yazdırma
print("\nKüme Merkezleri (Ortalama Süre ve Maliyetler):")
print("---------------------------------------------")
print(pd.DataFrame(centers, columns=['Ort. Süre', 'Ort. Maliyet']))
```
